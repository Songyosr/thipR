# Connect to THIP with R

This notebook uses thipR and your existing THIP database credentials. If THIP_DB_* variables are not already configured, uncomment thip_setup() below and run it once inside THIP. Enter the password through its masked prompt, never as cell text. Setup saves a private user .Renviron file (plaintext, not encrypted). Later sessions with normal R startup only need thip_connect(). All queries below use synthetic values or metadata.


```r
.libPaths(c(path.expand("~/R/thipR-library"), .libPaths()))
library(thipR)
library(dplyr)
# One-time setup, only if credentials are not already configured:
# thip_setup()
con <- thip_connect()
```

```r
DBI::dbGetQuery(con, "SELECT 1 AS connection_ok")
thip_list_tables(con)
```

## Inspect fields
Choose a table from the catalogue. This retrieves column definitions only.


```r
thip_columns(con, "nhso2568")
```

## Try ordinary dplyr syntax
The rows here are generated numbers, not records from a patient table. Filtering and aggregation execute in ClickHouse.


```r
query <- tbl(con, dbplyr::sql("SELECT number AS n FROM numbers(5)")) |>
  filter(n < 3) |>
  summarise(total = sum(n, na.rm = TRUE), rows = n())
show_query(query)
collect(query)
```
